```python
import numpy as np
from config import Config
from monitor_agent.monitor_runtime import MonitorRuntime
from sklearn.metrics import precision_recall_fscore_support

X = np.load(Config.SYNTHETIC_DATA_DIR / "cold_chain_X.npy")
y = np.load(Config.SYNTHETIC_DATA_DIR / "cold_chain_y_clf.npy")
np.random.seed(42)
idx = np.random.choice(len(X), 70000, replace=False)
test = idx[63000:70000][:400]            # 400 windows from the disjoint test split

m = MonitorRuntime()
probs = np.array([m.predict_anomaly(X[i])[30]["anomaly_probability"] for i in test])
true  = y[test]
p, r, f, _ = precision_recall_fscore_support(true, probs >= 0.80, average="binary")
print(f"P={p:.3f} R={r:.3f} F1={f:.3f}   (evaluated model: 0.800 / 0.802 / 0.801)")
print(f"mean prob on normal windows  : {probs[true==0].mean():.3f}")
print(f"mean prob on anomalous windows: {probs[true==1].mean():.3f}")
```
